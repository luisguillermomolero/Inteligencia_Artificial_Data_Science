# **Clase 1: Ética y Responsabilidad en IA**

**Objetivo general:**
Que los estudiantes comprendan los principios éticos en el desarrollo y despliegue de modelos de IA, identifiquen riesgos de sesgo y privacidad, y sean capaces de diseñar estrategias para mitigarlos, garantizando transparencia y responsabilidad en proyectos reales de IA.

---


## **1. Introducción a la ética en IA**

### **Teoría avanzada**

La ética en IA se centra en garantizar que los sistemas automatizados **respeten los derechos humanos, sean justos, confiables y transparentes**. Se fundamenta en principios internacionales:

* **Beneficencia:** La IA debe maximizar beneficios, promoviendo bienestar social y económico.
* **No maleficencia:** Prevenir daños directos o indirectos, incluyendo impactos psicológicos, sociales o financieros.
* **Justicia:** Garantizar equidad, evitando discriminación y asegurando igualdad de oportunidades.
* **Autonomía:** Permitir que los usuarios tomen decisiones informadas y comprendan cómo la IA impacta su entorno.
* **Transparencia y explicabilidad:** Hacer visibles los criterios de decisión de los modelos, facilitando la rendición de cuentas.




### **Normas y estándares internacionales**

* **UNESCO (2021):** Ética de IA y derechos humanos.
* **OCDE (2019):** Principios de IA confiable.
* **IEEE 7000:** Integración de valores éticos en el diseño de sistemas.
* **ISO/IEC 23053:** Frameworks de sistemas de IA.
* **ISO/IEC 23894:** Gestión de riesgos y auditoría de IA.



### **Ejemplos prácticos**

1. **COMPAS:** Predicción de reincidencia judicial con sesgos raciales. Los estudiantes analizan los impactos legales y sociales.
2. **Amazon Hiring Tool:** Sistema de selección que discriminaba candidatos por género, enseñando cómo los sesgos pueden perpetuarse en modelos históricos.
3. **Clearview AI:** Uso no autorizado de datos biométricos y problemas de privacidad, mostrando riesgos legales y éticos.



### **Actividad interactiva**

* **Debate rápido:** ¿Qué principios éticos se violaron en cada caso?
* Cada estudiante identifica un riesgo ético en un sistema que haya usado o conocido.

## **2. Sesgos en IA y su impacto**

### **Teoría**

Los sesgos pueden aparecer en distintas fases del ciclo de vida de IA:

* **Sesgo de datos:** Subrepresentación de minorías, etiquetado histórico discriminatorio, datos incompletos.
* **Sesgo de modelo:** Sobreajuste a patrones incorrectos, dependencia de variables sensibles.
* **Sesgo de uso:** Interpretación errónea de predicciones, decisiones sin supervisión humana.



**Consecuencias:**

* Decisiones injustas en salud, finanzas, empleo, justicia y vigilancia.
* Riesgo legal y reputacional.
* Desconfianza del usuario y rechazo de la tecnología.



### **Ejemplo detallado**

Dataset ficticio de créditos con variables sensibles: género, raza, edad, ubicación.

* Los estudiantes calculan métricas como **Disparate Impact** y **Demographic Parity** para detectar desigualdad.
* Observan cómo un modelo aparentemente “preciso” puede ser injusto para ciertos grupos.



### **Actividad práctica (mini-lab)**

* **Objetivo:** Detectar y corregir sesgos en un dataset.
* **Herramientas:** Python + pandas + scikit-learn.
* **Pasos:**

  1. Analizar distribución de datos por género y raza.
  2. Entrenar un modelo simple de clasificación de crédito.
  3. Evaluar métricas de fairness.
  4. Proponer estrategias de mitigación: re-muestreo, fairness constraints.
* **Discusión:** ¿Cómo se vería afectado el negocio y la sociedad si no se corrige el sesgo?

## **3. Privacidad y gobernanza de datos**

### **Teoría**

* **Privacidad:** Protección de datos personales y sensibles.
* **Regulaciones internacionales:** GDPR, ISO/IEC 27701, leyes locales sobre protección de datos.
* **Técnicas:**

  * **Anonimización:** Eliminación de identificadores directos.
  * **Pseudonimización:** Reemplazo de identificadores por códigos.
  * **Differential Privacy:** Añadir ruido para proteger individuos mientras se preserva utilidad.



### **Ejercicio práctico**

* Tomar un dataset de ejemplo y aplicar anonimización y pseudonimización.
* Analizar **riesgos residuales:** qué información aún podría revelar identidad.
* Reflexión grupal: ¿Qué equilibrio entre privacidad y utilidad de datos es aceptable en distintos contextos?

## **4. Transparencia y explicabilidad (XAI)**

### **Teoría avanzada**

* **Interpretabilidad:** Capacidad de entender decisiones de modelos simples (reglas, árboles de decisión).
* **Explicabilidad:** Herramientas que permiten comprender decisiones de modelos complejos (redes neuronales, ensembles).
* **Importancia:** Facilita auditorías, confianza del usuario y cumplimiento regulatorio.



### **Herramientas**

* **LIME:** Explicación local de predicciones.
* **SHAP:** Importancia de cada variable en la predicción.
* **Model Cards y Datasheets:** Documentos estandarizados que describen limitaciones, riesgos y uso correcto del modelo.



### **Actividad práctica**

* Mini-lab: explicar un modelo de clasificación de crédito o salud usando LIME o SHAP.
* Crear un **Model Card resumido**, describiendo riesgos, limitaciones y recomendaciones de uso responsable.

## **5. Auditoría ética de un modelo de IA**

### **Teoría**

* La auditoría ética consiste en evaluar sistemáticamente un modelo para identificar:

  * Sesgos y fairness.
  * Cumplimiento de privacidad.
  * Transparencia y explicabilidad.
  * Documentación y uso responsable.



### **Actividad grupal**

* Se entrega un **modelo de ejemplo (ficticio)** y un dataset de práctica.
* Cada grupo diseña un plan de auditoría:

  1. Identificación de riesgos éticos.
  2. Estrategias de mitigación.
  3. Creación de un Model Card resumido.
* Presentación rápida de hallazgos y feedback del instructor.
