# Introducción a MLOps 

### ¿Qué es MLOps?

- Cuando trabajamos en **Machine Learning**, solemos entrenar modelos en notebooks como Jupyter o Google Colab. 

- Allí todo funciona bien porque el entorno está controlado.

- Pero en una empresa, ese modelo debe funcionar con **miles de usuarios, datos nuevos todos los días y bajo estándares de calidad**, no solo en tu máquina.

Ahí es donde entra **MLOps**. **MLOps es la combinación de Machine Learning + DevOps.**. DevOps se enfoca en automatizar el despliegue de software.

**MLOps lleva esa idea al mundo de la IA**, incluyendo también:

✔ versionamiento de datos y modelos

✔ control de experimentos

✔ despliegue automatizado

✔ monitoreo continuo

✔ capacidad de volver atrás si algo falla

✔ reentrenamiento automático cuando los datos cambian

🔹 **En resumen:**
*MLOps es un conjunto de prácticas, herramientas y procesos que permiten llevar un modelo de IA desde la fase de desarrollo hasta su despliegue en producción de forma confiable, automática y escalable.*

---



### ¿Por qué es necesario MLOps?

Porque entrenar un modelo en clase o laboratorio es fácil.
Lo difícil es que **funcione correctamente fuera de ese entorno**, en una aplicación real, sin errores y de forma sostenida en el tiempo.

Imagina este escenario:

> Un modelo de predicción de fraude funciona muy bien con datos del año pasado. Lo pones en producción hoy, pero dentro de tres meses los patrones de fraude cambian. Si nadie lo monitorea ni reentrena, el modelo empieza a fallar… silenciosamente.

☠️ Resultado: predicciones incorrectas → decisiones erradas → pérdidas económicas → riesgo reputacional.

---


### 🧨 Principales problemas que MLOps resuelve

#### 1. *Modelos que funcionan en notebooks pero fallan en producción.*

En los notebooks el entorno es controlado, los datos son pequeños y el código no está optimizado.
En producción cambian los servidores, librerías, formatos de datos y el modelo se rompe.

**Solución MLOps:** contenedores (Docker), automatización de despliegue, pruebas automáticas.

---


#### 2. *No se pueden replicar los experimentos.*

Un estudiante prueba un modelo con cierto dataset y obtiene 92% de precisión. Un compañero intenta repetirlo y obtiene 85%. ¿Por qué? → No se versionaron los datos, parámetros ni código correctamente.

**Solución MLOps:** uso de herramientas como MLflow o DVC para registrar todo lo que se hizo (dataset, parámetros, librerías, métricas).

---


#### 3. *El modelo empeora con el tiempo: “data drift”.*

Los modelos se entrenan con datos históricos. Si cambian los hábitos de los usuarios o el entorno del negocio, **el modelo empieza a tomar malas decisiones**.

**Solución MLOps:** monitoreo en tiempo real y reentrenamiento automatizado cuando el rendimiento baja.

---


#### 4. *Falta de integración con otros equipos (IT, negocio, seguridad).*

El científico de datos entrena el modelo, pero:

* TI debe integrarlo con sistemas empresariales,
* Seguridad debe garantizar protección de datos,
* El área de negocio debe validar la utilidad.

Si no hay colaboración, el proyecto puede fallar aunque el modelo sea técnicamente correcto.

**Solución MLOps:** adopción de pipelines automatizados, documentación clara y herramientas colaborativas.

---


### 📌 Resumen 

👉 *MLOps es lo que permite que un modelo de IA deje de ser un ejercicio académico y se convierta en una solución real para una empresa.*

📊 Con MLOps se logra que los modelos:
✔ se entrenen correctamente,
✔ se desplieguen de forma automática,
✔ se mantengan actualizados,
✔ sean monitorizados continuamente,
✔ y se puedan reproducir cuantas veces sea necesario.

# 6. Fase 1 – Ingesta y validación de datos

Objetivo:

* Leer el CSV original.
* Verificar que tenga todas las columnas esperadas.
* Comprobar que no haya demasiados valores nulos.
* Guardar un archivo “validado” como base para el resto del pipeline.

Archivo: `ingest_and_validate.py`



````python
"""
Fase 1: Ingesta y validación de datos.

En esta etapa simulamos una validación mínima:
- Que existan todas las columnas esperadas.
- Que no haya columnas con demasiados valores nulos.

En un MLOps real, esta validación podría ser orquestada por Airflow
y apoyarse en herramientas de validación de datos como Great Expectations.
"""

# Importamos la librería pandas, muy usada para manejar datos en forma de tablas (DataFrames)
import pandas as pd

# Importamos Path desde pathlib, que nos permite trabajar con rutas de archivos de forma más cómoda y segura
from pathlib import Path

# -------------------------------------------------------------------
# CONFIGURACIÓN BÁSICA DEL SCRIPT
# -------------------------------------------------------------------

# Ruta del dataset original.
# Path crea un objeto "ruta" independiente del sistema operativo (Windows, Linux, etc.)
DATA_PATH = Path("data/churn_data.csv")

# Lista con los nombres de las columnas que esperamos encontrar en el archivo CSV.
# Esto define el "esquema" mínimo del dataset.
EXPECTED_COLUMNS = [
    "age",
    "tenure",
    "monthly_fee",
    "num_products",
    "has_partner",
    "has_dependents",
    "churn",
]

# Definimos el porcentaje máximo de valores nulos permitido por columna.
# Por ejemplo, 0.3 equivale al 30% de valores nulos como máximo.
MAX_NULL_RATIO = 0.3


# -------------------------------------------------------------------
# FUNCIÓN 1: INGESTA DE DATOS
# -------------------------------------------------------------------
def ingest_data(path: Path) -> pd.DataFrame:
    """
    Carga el dataset desde disco.

    Parámetros:
        path (Path): ruta al archivo CSV.

    Retorna:
        pd.DataFrame: datos cargados en un DataFrame de pandas.
    """
    # Verificamos si el archivo existe en la ruta indicada.
    if not path.exists():
        # Si el archivo no existe, lanzamos un error explícito con un mensaje claro.
        raise FileNotFoundError(f"No se encontró el archivo: {path}")
    
    # Si el archivo existe, lo leemos usando pandas.read_csv y lo guardamos en df.
    df = pd.read_csv(path)
    
    # Imprimimos un mensaje indicando que la carga fue exitosa y mostramos el "shape" (filas, columnas).
    print(f"[CARGA] Datos cargados con shape: {df.shape}")
    
    # Devolvemos el DataFrame para que pueda usarse en otras funciones.
    return df


# -------------------------------------------------------------------
# FUNCIÓN 2: VALIDACIÓN DEL ESQUEMA
# -------------------------------------------------------------------
def validate_schema(df: pd.DataFrame) -> None:
    """
    Valida que existan todas las columnas esperadas en el DataFrame.

    Parámetros:
        df (pd.DataFrame): DataFrame con los datos cargados.

    No retorna nada, pero lanza un error si faltan columnas.
    """
    # Recorremos EXPECTED_COLUMNS y nos quedamos con aquellas que NO están en df.columns.
    # Es decir, construimos una lista de columnas faltantes.
    missing = [c for c in EXPECTED_COLUMNS if c not in df.columns]
    
    # Si la lista de columnas faltantes NO está vacía, significa que el esquema es incorrecto.
    if missing:
        # Lanzamos un ValueError con el detalle de cuáles columnas faltan.
        raise ValueError(f"Faltan columnas en el dataset: {missing}")
    
    # Si no faltan columnas, imprimimos un mensaje de éxito.
    print("[OK] Esquema validado (todas las columnas estan presentes).")


# -------------------------------------------------------------------
# FUNCIÓN 3: VALIDACIÓN DE NULOS
# -------------------------------------------------------------------
def validate_nulls(df: pd.DataFrame) -> None:
    """
    Valida que no haya columnas con un porcentaje de valores nulos
    mayor al umbral definido en MAX_NULL_RATIO.

    Parámetros:
        df (pd.DataFrame): DataFrame con los datos cargados.

    No retorna nada, pero lanza un error si se encuentran columnas problemáticas.
    """
    # Calculamos el porcentaje de valores nulos por columna usando df.isnull().mean().
    # Esto devuelve una serie donde cada valor está entre 0 y 1 (0% a 100%).
    null_ratios = df.isnull().mean()
    
    # Filtramos aquellas columnas cuyo porcentaje de nulos sea mayor al límite permitido.
    problematic = null_ratios[null_ratios > MAX_NULL_RATIO]
    
    # Si 'problematic' no está vacío, significa que hay columnas con demasiados nulos.
    if not problematic.empty:
        # Lanzamos un ValueError mostrando cuáles columnas tienen demasiados nulos y qué porcentaje.
        raise ValueError(
            "Hay columnas con demasiados valores nulos:\n"
            f"{problematic}"
        )
    
    # Si no hay columnas problemáticas, mostramos un mensaje de éxito.
    print("[OK] Porcentaje de nulos aceptable en todas las columnas.")


# -------------------------------------------------------------------
# FUNCIÓN PRINCIPAL: ORQUESTA LA FASE 1
# -------------------------------------------------------------------
def main() -> None:
    """
    Función principal del script.

    Orquesta el flujo:
    1. Cargar datos desde el CSV.
    2. Validar el esquema.
    3. Validar los valores nulos.
    4. Guardar una versión validada del dataset.
    """
    # 1) Ingestamos los datos desde la ruta DATA_PATH.
    df = ingest_data(DATA_PATH)
    
    # 2) Validamos que el esquema (columnas) sea el esperado.
    validate_schema(df)
    
    # 3) Validamos que el porcentaje de nulos sea aceptable.
    validate_nulls(df)

    # 4) Si todo salió bien, definimos la ruta de salida para los datos validados.
    output_path = Path("data/churn_validated.csv")
    
    # Guardamos el DataFrame 'df' en un nuevo archivo CSV SIN índice.
    df.to_csv(output_path, index=False)
    
    # Imprimimos un mensaje final indicando dónde se guardaron los datos validados.
    print(f"[GUARDADO] Datos validados guardados en: {output_path}")


# -------------------------------------------------------------------
# PUNTO DE ENTRADA DEL SCRIPT
# -------------------------------------------------------------------
# Este bloque se ejecuta solo si el archivo se corre directamente con:
# python ingest_and_validate.py
# y NO cuando se importa como módulo desde otro script.
if __name__ == "__main__":
    # Llamamos a la función principal para ejecutar toda la fase de ingesta y validación.
    main()
````

Esta etapa enlaza con MLOps porque:

* En un entorno profesional, se automatiza esta validación diariamente.
* Airflow/Kubeflow podrían lanzar esta tarea como primer paso del pipeline.

# 7. Fase 2 – Preprocesamiento y división train/test

Objetivo:

* Separar características (features) y etiqueta (target).
* Escalar las variables numéricas.
* Dividir en train/test.
* Guardar:

  * `data/train.csv`
  * `data/test.csv`
  * `models/preprocessor.pkl` (escalador)

Archivo: `preprocess_and_split.py`



````python
"""
Fase 2: Preprocesamiento y división train/test.

Aquí:
- Seleccionamos las columnas de entrada (features) y la columna objetivo.
- Escalamos las características numéricas con StandardScaler.
- Dividimos en train/test.
- Guardamos los datasets ya transformados y el preprocesador.

En MLOps, esta fase también se versiona (por ejemplo con DVC),
para saber con qué datos exactos entrenamos cada modelo.
"""

# Importamos pandas para manipular los datos en forma de DataFrame
import pandas as pd

# Importamos Path para manejar rutas de carpetas/archivos de forma segura
from pathlib import Path

# Función para dividir datos en entrenamiento y prueba de forma estadísticamente válida
from sklearn.model_selection import train_test_split

# StandardScaler se usa para normalizar/escalar datos numéricos
from sklearn.preprocessing import StandardScaler

# Pickle permite guardar objetos de Python (como el preprocesador) en disco
import pickle

# --------------------------------------------------------------------
# DEFINICIÓN DE RUTAS DE ARCHIVOS
# --------------------------------------------------------------------

# Archivo generado en la Fase 1 con datos validados
DATA_PATH = Path("data/churn_validated.csv")

# Rutas donde guardaremos los datasets procesados
TRAIN_PATH = Path("data/train.csv")
TEST_PATH = Path("data/test.csv")

# Ruta donde se guardará el preprocesador (escalador)
PREPROCESSOR_PATH = Path("models/preprocessor.pkl")

# --------------------------------------------------------------------
# DEFINICIÓN DE COLUMNAS
# --------------------------------------------------------------------

# Variables de entrada (features) que el modelo usará para aprender
FEATURE_COLS = [
    "age",
    "tenure",
    "monthly_fee",
    "num_products",
    "has_partner",
    "has_dependents",
]

# Variable objetivo: lo que queremos predecir (cliente se va o no)
TARGET_COL = "churn"


# --------------------------------------------------------------------
# FUNCIÓN PRINCIPAL
# --------------------------------------------------------------------
def main() -> None:
    # 1️⃣ CARGAMOS LOS DATOS VALIDADOS
    df = pd.read_csv(DATA_PATH)
    print(f"[INFO] Datos validados cargados: {df.shape}")  # shape → (n_filas, n_columnas)

    # Separamos features (X) y target (y)
    X = df[FEATURE_COLS]      # Variables independientes
    y = df[TARGET_COL]        # Variable dependiente

    # 2️⃣ ESCALADO DE LOS DATOS
    # Creamos el escalador e identificamos estadísticas con fit_transform
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    # 3️⃣ DIVISIÓN ENTRE TRAIN Y TEST
    X_train, X_test, y_train, y_test = train_test_split(
        X_scaled,     # Usamos datos ya escalados
        y,
        test_size=0.2,     # 20% datos para prueba
        random_state=42,   # Asegura reproducibilidad
        stratify=y         # Mantiene proporción de clases (0/1)
    )

    # 4️⃣ RECONSTRUCCIÓN DE DATAFRAMES
    # X_train y X_test están ahora como arrays → los volvemos DataFrame
    train_df = pd.DataFrame(X_train, columns=FEATURE_COLS)
    train_df[TARGET_COL] = y_train.reset_index(drop=True)  # Agregamos la columna objetivo

    test_df = pd.DataFrame(X_test, columns=FEATURE_COLS)
    test_df[TARGET_COL] = y_test.reset_index(drop=True)

    # 5️⃣ GUARDAMOS LOS ARCHIVOS RESULTANTES
    TRAIN_PATH.parent.mkdir(parents=True, exist_ok=True)  # Crea carpeta si no existe
    train_df.to_csv(TRAIN_PATH, index=False)              # Guardamos training set

    test_df.to_csv(TEST_PATH, index=False)                # Guardamos test set

    # 6️⃣ GUARDAMOS EL ESCALADOR PARA REUTILIZAR EN PRODUCCIÓN
    PREPROCESSOR_PATH.parent.mkdir(parents=True, exist_ok=True)
    with open(PREPROCESSOR_PATH, "wb") as f:
        pickle.dump(scaler, f)  # Guardamos StandardScaler

    # Mensajes finales para verificación
    print(f"[GUARDADO] Train guardado en: {TRAIN_PATH}")
    print(f"[GUARDADO] Test guardado en: {TEST_PATH}")
    print(f"[GUARDADO] Preprocesador guardado en: {PREPROCESSOR_PATH}")


# Punto de entrada del script
if __name__ == "__main__":
    main()
````

# 8. Fase 3 – Entrenamiento del modelo con MLflow (tracking de experimentos)

Objetivo:

* Entrenar un modelo de clasificación (LogisticRegression).
* Registrar parámetros y métricas con MLflow.
* Guardar el modelo entrenado en `models/model.pkl`.

Archivo: `train.py`



````python
"""
Fase 3: Entrenamiento del modelo con MLflow.

Aquí:
- Cargamos el dataset de entrenamiento ya preprocesado.
- Entrenamos un modelo de Regresión Logística.
- Registramos métricas y parámetros en MLflow.
- Guardamos el modelo en disco.

Esta fase, en MLOps, suele correr en un job desatendido
(lanzado por Jenkins, GitHub Actions, Airflow, etc.).
"""

# -----------------------------------------------------------
# IMPORTACIONES NECESARIAS
# -----------------------------------------------------------

# pandas para manipular DataFrames
import pandas as pd

# Path para trabajar con rutas cruzadas plataforma
from pathlib import Path

# Modelo de clasificación
from sklearn.linear_model import LogisticRegression

# Métricas de evaluación
from sklearn.metrics import accuracy_score, f1_score

# Librería para tracking de experimentos
import mlflow
import mlflow.sklearn

# Para registrar la estructura de entrada/salida del modelo
from mlflow.models import infer_signature

# Para guardar el modelo localmente
import pickle


# -----------------------------------------------------------
# CONFIGURACIÓN DE MLflow
# -----------------------------------------------------------

# Indicamos que MLflow usará SQLite como base de datos para tracking.
# Esto evita el uso de filesystem plano y previene warnings.
mlflow.set_tracking_uri("sqlite:///mlflow.db")


# -----------------------------------------------------------
# RUTAS Y CONFIGURACIÓN DE COLUMNAS
# -----------------------------------------------------------

# Ruta al conjunto de entrenamiento preparado en la fase 2
TRAIN_PATH = Path("data/train.csv")

# Ruta donde guardaremos el modelo entrenado localmente
MODEL_PATH = Path("models/model.pkl")

# Columnas con las variables de entrada
FEATURE_COLS = [
    "age",
    "tenure",
    "monthly_fee",
    "num_products",
    "has_partner",
    "has_dependents",
]

# Variable objetivo (lo que queremos predecir)
TARGET_COL = "churn"


# -----------------------------------------------------------
# FUNCIÓN PRINCIPAL DEL SCRIPT
# -----------------------------------------------------------
def main() -> None:
    # 1️⃣ CARGA DE DATOS DE ENTRENAMIENTO
    df_train = pd.read_csv(TRAIN_PATH)

    # Separamos las features y la variable objetivo
    X_train = df_train[FEATURE_COLS]
    y_train = df_train[TARGET_COL]

    # 2️⃣ DEFINIMOS EL EXPERIMENTO EN MLflow
    # Si no existe, MLflow lo crea automáticamente
    mlflow.set_experiment("churn_mlops_demo")

    # 3️⃣ INICIAMOS UNA EJECUCIÓN (RUN)
    # Cada run representa una ejecución de entrenamiento
    with mlflow.start_run(run_name="training"):

        # Definimos los hiperparámetros del modelo
        # En producción pueden venir de un YAML, API o AutoML
        params = {
            "C": 1.0,
            "max_iter": 200,
            "solver": "liblinear",
        }

        # Creamos el modelo con los hiperparámetros definidos
        model = LogisticRegression(
            C=params["C"],
            max_iter=params["max_iter"],
            solver=params["solver"],
        )

        # 4️⃣ ENTRENAMOS EL MODELO
        model.fit(X_train, y_train)

        # 5️⃣ REALIZAMOS PREDICCIONES SOBRE TRAIN (SOLO PARA VALIDACIÓN INICIAL)
        y_pred = model.predict(X_train)

        # Calculamos métricas de rendimiento
        acc = accuracy_score(y_train, y_pred)
        f1 = f1_score(y_train, y_pred)

        # Imprimimos resultados por consola
        print(f"[OK] Accuracy (train): {acc:.4f}")
        print(f"[OK] F1-score (train): {f1:.4f}")

        # 6️⃣ REGISTRAMOS MÉTRICAS Y PARÁMETROS EN MLflow
        mlflow.log_params(params)
        mlflow.log_metric("train_accuracy", acc)
        mlflow.log_metric("train_f1", f1)

        # -----------------------------------------------------------
        # 7️⃣ GUARDAMOS EL MODELO LOCALMENTE (PICKLE)
        # -----------------------------------------------------------
        MODEL_PATH.parent.mkdir(parents=True, exist_ok=True)
        with open(MODEL_PATH, "wb") as f:
            pickle.dump(model, f)

        # -----------------------------------------------------------
        # 8️⃣ REGISTRO COMPLETO DEL MODELO EN MLflow
        # -----------------------------------------------------------
        # Creamos una "signature" que describe inputs y outputs
        signature = infer_signature(X_train, y_pred)

        # Tomamos una fila de ejemplo como input_example
        input_example = X_train.head(1)

        # Registramos el modelo en MLflow con metadata
        mlflow.sklearn.log_model(
            sk_model=model,
            name="model",
            signature=signature,
            input_example=input_example,
            registered_model_name="churn_model"  # Nombre dentro del Model Registry
        )

        print(f"[GUARDADO] Modelo guardado en: {MODEL_PATH}")
        print("[INFO] Modelo registrado en MLflow.")


# -----------------------------------------------------------
# PUNTO DE ENTRADA
# -----------------------------------------------------------
if __name__ == "__main__":
    main()
````

# 9. Fase 4 – Evaluación del modelo

Objetivo:

* Evaluar el rendimiento en el conjunto de prueba.
* Registrar métricas en MLflow (test).

Archivo: `evaluate.py`



````python
"""
Fase 4: Evaluación del modelo.

Aquí:
- Cargamos el modelo entrenado.
- Evaluamos en el conjunto de prueba.
- Registramos métricas en MLflow.

Esta evaluación es clave en MLOps para decidir si un modelo
es lo suficientemente bueno para ir a producción.
"""

# Importamos pandas para manejar el dataset de test
import pandas as pd

# Importamos Path para manejar rutas de forma segura
from pathlib import Path

# Importamos métricas para evaluar rendimiento
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score

# Importamos pickle para cargar el modelo entrenado desde disco
import pickle

# Importamos MLflow para registrar los resultados de la evaluación
import mlflow

# -------------------------------------------------------------------
# CONFIGURACIÓN DE MLflow
# -------------------------------------------------------------------

# Establecemos que MLflow usará una base de datos SQLite como almacenamiento
# Esto permite tener un backend persistente más adecuado que solo archivos locales
mlflow.set_tracking_uri("sqlite:///mlflow.db")


# -------------------------------------------------------------------
# RUTAS Y DEFINICIÓN DE COLUMNAS
# -------------------------------------------------------------------

# Ruta del archivo CSV con los datos de prueba (test set)
TEST_PATH = Path("data/test.csv")

# Ruta donde se encuentra el modelo entrenado
MODEL_PATH = Path("models/model.pkl")

# Columnas de entrada (features)
FEATURE_COLS = [
    "age",
    "tenure",
    "monthly_fee",
    "num_products",
    "has_partner",
    "has_dependents",
]

# Nombre de la columna objetivo (target)
TARGET_COL = "churn"


# -------------------------------------------------------------------
# FUNCIÓN PRINCIPAL DEL SCRIPT
# -------------------------------------------------------------------
def main() -> None:
    """
    Fase de evaluación del modelo:
    1. Se cargan datos de prueba.
    2. Se carga el modelo entrenado.
    3. Se generan predicciones.
    4. Se calculan métricas.
    5. Se registran en MLflow.
    """

    # 1) Cargamos el dataset de prueba ya escalado y con la columna objetivo
    df_test = pd.read_csv(TEST_PATH)
    
    # Separamos features y target para la evaluación
    X_test = df_test[FEATURE_COLS]
    y_test = df_test[TARGET_COL]


    # 2) Cargamos el modelo entrenado desde disco usando pickle
    with open(MODEL_PATH, "rb") as f:
        model = pickle.load(f)


    # 3) Generamos las predicciones del modelo
    # y_pred → predicción binaria (0/1)
    # y_proba → probabilidad de churn (solo columna correspondiente a clase positiva)
    y_pred = model.predict(X_test)
    y_proba = model.predict_proba(X_test)[:, 1]


    # 4) Calculamos métricas de evaluación
    # Accuracy → porcentaje de aciertos
    acc = accuracy_score(y_test, y_pred)

    # F1-score → balance entre precisión y sensibilidad
    f1 = f1_score(y_test, y_pred)

    # ROC-AUC → mide calidad del modelo basándose en probabilidad
    roc = roc_auc_score(y_test, y_proba)

    # Mostramos resultados por consola
    print(f"[OK] Test Accuracy: {acc:.4f}")
    print(f"[OK] Test F1-score: {f1:.4f}")
    print(f"[OK] Test ROC-AUC:  {roc:.4f}")


    # 5) Registramos resultados en MLflow
    # Esto permite comparar modelos entrenados en diferentes momentos
    mlflow.set_experiment("churn_mlops_demo")

    # Iniciamos una ejecución (run) específica para la evaluación
    with mlflow.start_run(run_name="evaluation"):
        mlflow.log_metric("test_accuracy", acc)
        mlflow.log_metric("test_f1", f1)
        mlflow.log_metric("test_roc_auc", roc)


# -------------------------------------------------------------------
# PUNTO DE ENTRADA DEL SCRIPT
# -------------------------------------------------------------------
if __name__ == "__main__":
    main()
````

# 10. Fase 5 – Monitorización y reentrenamiento automático

Objetivo:

* Simular llegada de datos nuevos (`new_churn_data.csv`).
* Evaluar el modelo actual en esos datos.
* Si el rendimiento cae por debajo de un umbral, reentrenar.

Archivo: `monitor_and_retrain.py`



````python
"""
Fase 5: Monitorización y reentrenamiento.

Simulamos:
- Nuevos datos que llegan (new_churn_data.csv).
- Evaluamos el modelo actual en esos datos.
- Si el accuracy es menor a un umbral, reentrenamos el modelo
  con los datos antiguos + los nuevos.

En MLOps real, esta tarea se programa (cron, Airflow, Jenkins)
y se ejecuta de manera periódica (por ejemplo, cada noche).
"""

# -------------------------------------------------------------------
# IMPORTACIONES
# -------------------------------------------------------------------

# Pandas para manejo tabular de datos
import pandas as pd

# Path para rutas seguras en sistemas operativos
from pathlib import Path

# Modelo base y función de evaluación
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Para dividir datos en entrenamiento y validación
from sklearn.model_selection import train_test_split

# Para registrar estructura de entrada/salida del modelo en MLflow
from mlflow.models import infer_signature

# MLflow para tracking
import mlflow
import mlflow.sklearn

# pickle para cargar y guardar modelos y preprocesadores
import pickle


# -------------------------------------------------------------------
# CONFIGURACIÓN DE MLflow
# -------------------------------------------------------------------

# Usamos SQLite como backend de tracking para MLflow
mlflow.set_tracking_uri("sqlite:///mlflow.db")

# -------------------------------------------------------------------
# RUTAS Y PARÁMETROS DE CONFIGURACIÓN
# -------------------------------------------------------------------

# Nuevos datos que llegan (simulación de producción)
NEW_DATA_PATH = Path("data/new_churn_data.csv")

# Conjunto de entrenamiento histórico
TRAIN_PATH = Path("data/train.csv")

# Modelo actual y preprocesador entrenado
MODEL_PATH = Path("models/model.pkl")
PREPROCESSOR_PATH = Path("models/preprocessor.pkl")

# Columnas de entrada
FEATURE_COLS = [
    "age",
    "tenure",
    "monthly_fee",
    "num_products",
    "has_partner",
    "has_dependents",
]

# Columna objetivo
TARGET_COL = "churn"

# Umbral mínimo de rendimiento (decisión MLOps)
PERFORMANCE_THRESHOLD = 0.80  # 80% de accuracy


# -------------------------------------------------------------------
# FUNCIÓN PRINCIPAL
# -------------------------------------------------------------------
def main() -> None:
    """
    Simula la llegada de nuevos datos y evalúa el modelo.
    Si el rendimiento cae por debajo del umbral, se reentrena.
    """

    # 1) Verificamos si hay archivos nuevos
    if not NEW_DATA_PATH.exists():
        print("[ADVERTENCIA] No hay datos nuevos para monitorizar.")
        return


    # ----------------------------------------------------------------
    # 2) Carga de datos nuevos y preprocesador
    # ----------------------------------------------------------------

    # Cargamos los datos que simulan nueva información en producción
    df_new_raw = pd.read_csv(NEW_DATA_PATH)

    # Cargamos el preprocesador (StandardScaler) para transformar datos
    with open(PREPROCESSOR_PATH, "rb") as f:
        scaler = pickle.load(f)

    # Separamos features y target
    X_new = df_new_raw[FEATURE_COLS]
    y_new = df_new_raw[TARGET_COL]

    # Aplicamos el escalado
    X_new_scaled = scaler.transform(X_new)


    # ----------------------------------------------------------------
    # 3) Carga del modelo actual
    # ----------------------------------------------------------------
    with open(MODEL_PATH, "rb") as f:
        current_model = pickle.load(f)


    # ----------------------------------------------------------------
    # 4) Evaluación del modelo en datos nuevos
    # ----------------------------------------------------------------
    y_pred_new = current_model.predict(X_new_scaled)
    acc_new = accuracy_score(y_new, y_pred_new)

    print(f"[INFO] Accuracy en datos nuevos: {acc_new:.4f}")

    # Registramos esta evaluación en MLflow para seguimiento histórico
    mlflow.set_experiment("churn_mlops_demo")
    with mlflow.start_run(run_name="monitoring"):
        mlflow.log_metric("new_data_accuracy", acc_new)


    # ----------------------------------------------------------------
    # 5) ¿Reentrenamos? → SI rendimiento < umbral
    # ----------------------------------------------------------------
    if acc_new < PERFORMANCE_THRESHOLD:
        print("[ADVERTENCIA] Rendimiento bajo, iniciando reentrenamiento...")


        # ------------------------------------------------------------
        # 5.1 Carga de datos previos (históricos)
        # ------------------------------------------------------------
        df_train_scaled = pd.read_csv(TRAIN_PATH)

        # Formamos un DataFrame con datos nuevos escalados
        df_new_scaled = pd.DataFrame(X_new_scaled, columns=FEATURE_COLS)
        df_new_scaled[TARGET_COL] = y_new.reset_index(drop=True)

        # Combinamos datos antiguos + nuevos
        df_full = pd.concat(
            [df_train_scaled, df_new_scaled],
            ignore_index=True
        )

        X_full = df_full[FEATURE_COLS]
        y_full = df_full[TARGET_COL]


        # ------------------------------------------------------------
        # 5.2 División en entrenamiento + validación
        # ------------------------------------------------------------
        X_train, X_val, y_train, y_val = train_test_split(
            X_full,
            y_full,
            test_size=0.2,
            random_state=42,
            stratify=y_full
        )

        # ------------------------------------------------------------
        # 5.3 Reentrenamiento del modelo
        # ------------------------------------------------------------
        new_model = LogisticRegression(
            max_iter=300,
            solver="liblinear"
        )
        new_model.fit(X_train, y_train)

        # Evaluamos rendimiento del nuevo modelo
        acc_val = accuracy_score(y_val, new_model.predict(X_val))
        print(f"[OK] Nuevo modelo entrenado. Accuracy validación: {acc_val:.4f}")


        # ------------------------------------------------------------
        # 5.4 Guardar modelo actualizado
        # ------------------------------------------------------------
        with open(MODEL_PATH, "wb") as f:
            pickle.dump(new_model, f)

        # Registrar en MLflow
        with mlflow.start_run(run_name="retrain"):
            mlflow.log_metric("retrain_val_accuracy", acc_val)

            # Crear signature e input_example
            signature = infer_signature(X_train, new_model.predict(X_train))
            input_example = X_train.head(1)

            mlflow.sklearn.log_model(
                sk_model=new_model,
                name="model",
                signature=signature,
                input_example=input_example,
                registered_model_name="churn_model"
            )

        print("[GUARDADO] Modelo actualizado guardado en disco y registrado en MLflow.")

    else:
        print("[OK] El modelo mantiene un rendimiento aceptable. No se reentrena.")


# -------------------------------------------------------------------
# PUNTO DE ENTRADA
# -------------------------------------------------------------------
if __name__ == "__main__":
    main()
````

# 11. Fase 6 – API de predicción con FastAPI

Objetivo:

* Servir el modelo como un servicio HTTP.
* Permitir que otras aplicaciones llamen una URL y obtengan la probabilidad de churn.

Archivo: `serve_api.py`



````python
"""
Fase 6: Despliegue como API con FastAPI.

Aquí:
- Cargamos el modelo y el preprocesador.
- Definimos una API REST con un endpoint /predict.
- Cualquier sistema (web, móvil, backend) podría consumir esta API.

En MLOps, esta API se ejecuta dentro de un contenedor Docker y
se orquesta con Kubernetes para asegurar disponibilidad y escalabilidad.
"""

from fastapi import FastAPI
from pydantic import BaseModel
from typing import Literal
import pickle
from pathlib import Path
import uvicorn

MODEL_PATH = Path("models/model.pkl")
PREPROCESSOR_PATH = Path("models/preprocessor.pkl")

FEATURE_COLS = [
    "age",
    "tenure",
    "monthly_fee",
    "num_products",
    "has_partner",
    "has_dependents",
]

# Cargamos modelo y preprocesador al iniciar el servicio
with open(MODEL_PATH, "rb") as f:
    model = pickle.load(f)

with open(PREPROCESSOR_PATH, "rb") as f:
    scaler = pickle.load(f)


class CustomerData(BaseModel):
    age: float
    tenure: float
    monthly_fee: float
    num_products: float
    has_partner: Literal[0, 1]
    has_dependents: Literal[0, 1]


app = FastAPI(
    title="Churn Prediction API",
    description="API de ejemplo para predecir la probabilidad de churn de un cliente.",
    version="1.0.0",
)


@app.post("/predict")
def predict_churn(data: CustomerData):
    # Orden de las features debe coincidir con el utilizado en el entrenamiento
    X = [[
        data.age,
        data.tenure,
        data.monthly_fee,
        data.num_products,
        data.has_partner,
        data.has_dependents,
    ]]

    # Escalamos con el mismo preprocesador
    X_scaled = scaler.transform(X)

    proba_churn = model.predict_proba(X_scaled)[0, 1]
    prediction = int(proba_churn > 0.5)

    return {
        "churn_probability": float(proba_churn),
        "churn_prediction": prediction
    }


if __name__ == "__main__":
    uvicorn.run(app, host="0.0.0.0", port=8000)
````
